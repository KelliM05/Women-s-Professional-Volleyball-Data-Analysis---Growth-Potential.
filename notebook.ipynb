```r
#kills
spike_graph <- function(team){
	# Graph Labels
	op <- team$Opponent
	opp <- head(op, -2)
	opps <- tail(opp,-1)
	Opponents <- gsub("\\s+", " ",opps)
	Opponents <- trimws(Opponents) 
	
	# Sum
	kills <- team$Spike.Total
	yval <- as.numeric(kills)
	values <- yval[!is.na(yval)]
	count <- head(values,-2) 
	# Errors
	error <- team$Spike.Errors
	yval2 <- as.numeric(error)
	values2 <- yval2[!is.na(yval2)]
	count2 <- head(values2,-2)
	# Blocks
	blocks <- team$Spike.Blocks
	yval3 <- as.numeric(blocks)
	values3 <- yval3[!is.na(yval3)]
	count3 <- head(values3,-2)
	# Points
	points <- team$Spike.Points
	yval4 <- as.numeric(points)
	values4 <- yval4[!is.na(yval4)]
	count4 <- head(values4,-2)
	# Hitting%
	hit <- team$Hitting..
	yval5 <- as.numeric(gsub("%","",hit))
	values5 <- yval5[!is.na(yval5)]
	count5 <- head(values5,-2)
	
	counts <- rbind(count5, #hitting%
					count2, # errors
					count3, # blocks
					count4, # kills
					count) # total atttempts 

	#totals
	t <- barplot(abs(counts),
		names.arg = Opponents,
		las=2,
		cex.names= 1,
        col = c("orange",
				"red",
			   "yellow",
			   "skyblue",
				"green"),
        border = "white",
        main = c("Hitting Statistics"),
	   horiz=TRUE)
t
	
#text above bar
	max_x <- max(counts)
	if ( max_x < 5) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x =  i - 0.5,
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)    
		}}} else if ( max_x < 20) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x = 0 + i,
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)
	}}}else if ( max_x < 26) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x = 0 + (i*2),
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)
	}}}else if ( max_x < 55) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x = 0 + (2*i^2),
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)
	}}}else if ( max_x < 101) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x = 0 + (3*i^2),
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)
	}}}else {
for (i in 1:nrow(counts)) {
	#for (j in 1:ncol(counts)) {
  text(x = 0 + 8*i^2,       
       y = t,                
       labels = counts[i, ],   
       cex = 1, pos = 4)        
   
}}

	
# Legend
	legend("topright",
       legend = c("Hitting %",
				  "Total Errors",
				  "Blocks",
				  "Kills", 
				  "Total Attempts"),
       fill = c("orange", #Hitting%
				"red", # errors
			   "yellow", # blocks
			   "skyblue", # total kills
				"green"), # points
       bty = "n")
	
	}
suppressWarnings(spike_graph(atl_match_stats))
```

```r
#serve
serve_graph <- function(team){
	# Graph Labels
	op <- team$Opponent
	opp <- head(op, -2)
	opps <- tail(opp,-1)
	Opponents <- gsub("\\s+", " ",opps)
	Opponents <- trimws(Opponents) 
	
	 
	# Errors
	error <- team$Service.Error.s.
	yval2 <- as.numeric(error)
	values2 <- yval2[!is.na(yval2)]
	count1 <- head(values2,-2)
	# Aces
	blocks <- team$Service.Ace.s
	yval3 <- as.numeric(blocks)
	values3 <- yval3[!is.na(yval3)]
	count2 <- head(values3,-2)
	# eff
	eff <- team$Serve.Eff.
	yval <- as.numeric(gsub("%","",eff))
	values <- yval[!is.na(yval)]
	count3 <- head(values,-2)
	# total
	tot <- team$Total.Serve
	yval4 <- as.numeric(tot)
	values4 <- yval4[!is.na(yval4)]
	count4 <- head(values4,-2)
	
	counts <- rbind(count1, 
					count2, 
					count3,
				   count4)
	
	#totals
	t <- barplot(abs(counts),
		names.arg = Opponents,
		las=2,
		cex.names= 1,
        col = c("red",
				"orange",
			   "pink",
			   "green"),
        border = "white",
        main = "Serving Statistics",
	   horiz=TRUE)

t
	

#text above bar
	max_x <- max(counts)
	if ( max_x < 5) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x =  i - 0.5,
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)    
		}}} else if ( max_x < 15) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x = 0 + i,
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)
	}}}else if ( max_x < 55) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x = 0 + (i^2),
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)
	}}}else {
for (i in 1:nrow(counts)) {
	#for (j in 1:ncol(counts)) {
  text(x = 0 + 6*i^2,       
       y = t,                
       labels = counts[i, ],   
       cex = 1, pos = 4)        
   
}}

	
# Legend
	legend("topright",
       legend = c("Errors",
				  "Aces",
				  "Eff %",
				 "Total"),
       fill = c("red", 
				"orange", 
			   "pink",
			   "green"),
       bty = "n")
	
	}
suppressWarnings(serve_graph(atl_match_stats))
```

```r
#Reception
recieve_graph <- function(team){
	# Graph Labels
	op <- team$Opponent
	opp <- head(op, -2)
	opps <- tail(opp,-1)
	Opponents <- gsub("\\s+", " ",opps)
	Opponents <- trimws(Opponents) 
	
	 
	# Errors
	error <- team$Reception.Errors
	yval <- as.numeric(error)
	values <- yval[!is.na(yval)]
	count <- head(values,-2)
	# total
	total <- team$Reception.Total
	yval2 <- as.numeric(total)
	values2 <- yval2[!is.na(yval2)]
	count2 <- head(values2,-2)
	# pos%
	pos <- team$Reception.Pos.
	yval3 <- as.numeric(gsub("%","",pos))
	values3 <- yval3[!is.na(yval3)]
	count3 <- head(values3,-2)
	# perf%
	perf <- team$Reception.Perf.
	yval4 <- as.numeric(gsub("%","",perf))
	values4 <- yval4[!is.na(yval4)]
	count4 <- head(values4,-2)
	
	counts <- rbind(count, count4, count3, count2)
	
	#totals
	t <- barplot(abs(counts),
		names.arg = Opponents,
		las=2,
		cex.names= 1,
        col = c("red",
				"orange",
			   "yellow",
			   "green"),
        border = "white",
        main = "Reception Statistics",
	   horiz=TRUE)
t
	
#text above bar
	max_x <- max(counts)
	if ( max_x == 1) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x =  0.5,
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)    
		}}}else if ( max_x < 2) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x =  0 - 0.5,
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)    
		}}}else if ( max_x < 5) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x =  i - 0.5,
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)    
		}}} else if ( max_x < 15) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x = 0 + i*2,
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)
	}}}else {
for (i in 1:nrow(counts)) {
	#for (j in 1:ncol(counts)) {
  text(x = 0 + 6*i^2,       
       y = t,                
       labels = counts[i, ],   
       cex = 1, pos = 4)        
   
}}
	
# Legend
	legend("topright",
       legend = c("Errors", "Perf %", "Pos %", "Totals"),
       fill = c("red", 
				"orange", 
			   "yellow",
			   "green"),
       bty = "n")
	
	}
suppressWarnings(recieve_graph(atl_match_stats))
```

```r
# Blocks
block_graph <- function(team){
	# Graph Labels
	op <- team$Opponent
	opp <- head(op, -2)
	opps <- tail(opp,-1)
	Opponents <- gsub("\\s+", " ",opps)
	Opponents <- trimws(Opponents) 
	
	# Points
	points <- team$Block.Points
	yval <- as.numeric(points)
	values <- yval[!is.na(yval)]
	count <- head(values,-2)
	# Touches
	touches <- team$Block.Touches
	yval2 <- as.numeric(touches)
	values2 <- yval2[!is.na(yval2)]
	count2 <- head(values2,-2)
	
	counts <- rbind(count2, count)
	
	#totals
	t <- barplot(counts,
		names.arg = Opponents,
		las=2,
		cex.names= 1,
        col = c("skyblue",
				"violet"),
        border = "white",
        main = "Blocking Statistics",
	   horiz=TRUE)
	
	
#text above bar
max_x <- max(counts)

if ( max_x < 5) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x =  i - 0.5,
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)    
		}}} else if ( max_x < 15) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x = 0 + i,
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)
	}}}else {
		for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x = 0 + (2*(i^3)),       
				 y = t[j],                
				 labels = counts[i, j],   
				 cex = 1, pos = 4) 
		}}}
	
# Legend
	legend("topright",
       legend = c("Touches",
				  "Points"),
       fill = c("skyblue", 
				"violet"),
       bty = "n")
	
	}
suppressWarnings(block_graph(atl_match_stats))
```

```r
# Totals
totals_graph <- function(team){
	# Graph Labels
	op <- team$Opponent
	opp <- head(op, -2)
	opps <- tail(opp,-1)
	Opponents <- gsub("\\s+", " ",opps)
	Opponents <- trimws(Opponents) 
	
	 
	# Sets
	sets <- team$Sets
	yval4 <- as.numeric(sets)
	values4 <- yval4[!is.na(yval4)]
	count4 <- head(values4,-2)
	# Points
	points <- team$Total.Points
	yval <- as.numeric(points)
	values <- yval[!is.na(yval)]
	count <- head(values,-2)
	# BP
	bp <- team$Points.BP
	yval2 <- as.numeric(bp)
	values2 <- yval2[!is.na(yval2)]
	count2 <- head(values2,-2)
	# Bal
	bal <- team$Points.Bal
	yval3 <- as.numeric(bal)
	values3 <- yval3[!is.na(yval3)]
	count3 <- head(values3,-2)
	
	counts <- rbind(count4, count2, count3, count)
	
	#totals
	t <- barplot(counts,
		names.arg = Opponents,
		las=2,
		cex.names= 1,
        col = c("pink",
				"yellow",
				"skyblue",
			   "green"),
        border = "white",
        main = "Totals",
	   horiz=TRUE)
	t
	

#text above bar
	max_x <- max(counts)
	if ( max_x < 15) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x =  i*2 ,
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)    
		}}} else if ( max_x < 50) {
for (i in 1:nrow(counts)) {
	for (j in 1:ncol(counts)) { 
			text(x = 0 + (2*i^2),
				 labels = counts[i, j], 
				 y = t[j],
				 cex = 1, pos = 4)
	}}}else {
for (i in 1:nrow(counts)) {
	#for (j in 1:ncol(counts)) {
  text(x = 0 + 6*i^2,       
       y = t,                
       labels = counts[i, ],   
       cex = 1, pos = 4)        
   
}}
# Legend
	legend("topright",
       legend = c("Sets Played",
				  "BP",
				  "Bal",
				 "Points Scored"),
       fill = c("pink",
		   "yellow",
				"skyblue",
			   "green"),
       bty = "n")
	
	}
suppressWarnings(totals_graph(atl_match_stats))
```
